```python
import pandas as pd

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name, sheet_name="Cleaned Data")

print("Dataset Shape:", df.shape)
display(df.head())
```

```python
df.describe()
```

```python
department_analysis = df.groupby("department").agg(
    students=("student_id", "count"),
    average_attendance=("attendance", "mean"),
    average_python_marks=("python_marks", "mean"),
    average_ml_marks=("ml_marks", "mean"),
    average_marks=("average_marks", "mean"),
    average_projects=("projects_completed", "mean")
).round(2)

display(department_analysis)
```

```python
import matplotlib.pyplot as plt

department_analysis["average_marks"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Student Performance by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
performance_counts = df["performance"].value_counts()

print(performance_counts)
```

```python
plt.figure(figsize=(8, 5))

performance_counts.plot(
    kind="bar"
)

plt.title("Student Performance Distribution")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["attendance"],
    df["average_marks"]
)

plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
correlation = df["attendance"].corr(df["average_marks"])

print("Correlation between Attendance and Average Marks:", round(correlation, 2))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["projects_completed"], df["average_marks"])

plt.title("Projects Completed vs Average Marks")
plt.xlabel("Projects Completed")
plt.ylabel("Average Marks")
plt.grid(True)
plt.tight_layout()
plt.show()

print(
    "Correlation:",
    round(df["projects_completed"].corr(df["average_marks"]), 2)
)
```

```python
year_performance = df.groupby("year")["average_marks"].mean().round(2)

print(year_performance)

plt.figure(figsize=(7, 5))

year_performance.plot(kind="bar")

plt.title("Average Performance by Year")
plt.xlabel("Academic Year")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("Overall Average Marks:", round(df["average_marks"].mean(), 2))
print("Average Attendance:", round(df["attendance"].mean(), 2))
print("Average Python Marks:", round(df["python_marks"].mean(), 2))
print("Average ML Marks:", round(df["ml_marks"].mean(), 2))
print("Average Projects:", round(df["projects_completed"].mean(), 2))

print("\nHighest Department:")
print(department_analysis["average_marks"].idxmax(),
      department_analysis["average_marks"].max())

print("\nLowest Department:")
print(department_analysis["average_marks"].idxmin(),
      department_analysis["average_marks"].min())

print("\nAttendance Correlation:",
      round(df["attendance"].corr(df["average_marks"]), 2))

print("Projects Correlation:",
      round(df["projects_completed"].corr(df["average_marks"]), 2))
```
